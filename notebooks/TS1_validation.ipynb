# TS1 morphology

## Setup

```python
from mascaf import *
from swctools import SWCModel, FrustaSet, PointSet, plot_model
import logging

logging.basicConfig(level=logging.INFO)
import os

print("✅ Libraries imported successfully!")
```

## Load Mesh

```python
spine_idx = 1
mcf_qst = 0.5
mcf_mcst = 10
obj_name = f"TS{spine_idx}"
polylines_name = f"TS{spine_idx}_qst{mcf_qst}_mcst{mcf_mcst}"
mm = MeshManager(mesh_path=f"../data/mesh/processed/{obj_name}.obj")
raw_skeleton = SkeletonGraph.from_txt(
    f"../data/mcf_skeletons/{polylines_name}.polylines.txt"
)
raw_skeleton.prune_short_branches_inplace(min_length=10)
mm.visualize_mesh_3d(skel=raw_skeleton)
```

## Skeleton optimization

```python
do_skeleton_optimization = True

if do_skeleton_optimization:
    opts = SkeletonOptimizerOptions(
        max_iterations=5,
        step_size=5.0,
        smoothing_weight=0.1,
        verbose=True,
    )
    print("\nOptimizing skeleton (serial)...")
    optimizer = SkeletonOptimizer(raw_skeleton, mm.mesh, opts)
    optimized_skeleton = optimizer.optimize()
    skeleton = optimized_skeleton
    mm.visualize_mesh_3d(skel=[raw_skeleton, optimized_skeleton], skel_color=["crimson", "blue"])
else:  # no optimization
    skeleton = raw_skeleton
```

```python
max_edge_length = 200

swc_out_dir = f"../data/swc/current/{polylines_name}"

# check if directory exists, if not create it
if not os.path.exists(swc_out_dir):
    os.makedirs(swc_out_dir)

radius_strategy_list = [
    "equivalent_area",
    # "equivalent_perimeter",
    # "section_median",
    # "section_circle_fit",
    # "nearest_surface",
]
for radius_strategy in radius_strategy_list:
    print(f"Computing skeleton for radius_strategy={radius_strategy} ...", end="")
    morph = fit_morphology(
        mm.mesh,
        skeleton,
        options=FitOptions(
            max_edge_length=max_edge_length,
            radius_strategy=radius_strategy,
            snap_polylines_to_mesh=True,
        ),
    )
    # write swc to file
    morph.to_swc_file(f"{swc_out_dir}/TS{spine_idx}_s{max_edge_length}_{radius_strategy}.swc")
    # validation
    validator = Validation(mm, skeleton, morph)
    for account_for_overlaps in [False, True]:
        vol_result = validator.compare_volumes(account_for_overlaps=account_for_overlaps)
        area_result = validator.compare_surface_areas(account_for_overlaps=account_for_overlaps)
        print(
            f"Validation Results ({radius_strategy}, account_for_overlaps={account_for_overlaps}):"
        )
        print("\nVolume Comparison:")
        print(f"  Mesh volume:       {vol_result['mesh_volume']:.4f}")
        print(f"  Morphology volume: {vol_result['morphology_volume']:.4f}")
        print(f"  Ratio:             {vol_result['ratio']:.4f}")
        print(f"  Error:             {vol_result['error']:.4f}")
        print(f"  Relative error:    {vol_result['relative_error']:.2%}")
        print("\nSurface Area Comparison:")
        print(f"  Mesh area:         {area_result['mesh_area']:.4f}")
        print(f"  Morphology area:   {area_result['morphology_area']:.4f}")
        print(f"  Ratio:             {area_result['ratio']:.4f}")
        print(f"  Error:             {area_result['error']:.4f}")
        print(f"  Relative error:    {area_result['relative_error']:.2%}")
    print()
print("DONE")
```

```python
# plot using swctools
make_html = True

for radius_strategy in radius_strategy_list:
    swc_filepath = f"{swc_out_dir}/TS{spine_idx}_s{max_edge_length}_{radius_strategy}.swc"
    model = SWCModel.from_swc_file(swc_filepath)
    model.print_attributes(node_info=False, edge_info=False)
    frusta = FrustaSet.from_swc_model(model)
    title = f"TS{spine_idx}_s{max_edge_length}_{radius_strategy}"
    fig = plot_model(swc_model=model, frusta=frusta, slider=True, title=title)
    fig.show()
    if make_html:
        fig.write_html(f"../viz/TS{spine_idx}_s{max_edge_length}_{radius_strategy}.html")
```

## Normalize radii glabally to total surface area

```python
morph = fit_morphology(
    mm.mesh,
    skeleton,
    options=FitOptions(
        max_edge_length=max_edge_length,
        radius_strategy=radius_strategy,
        snap_polylines_to_mesh=False,
    ),
)

# Configure and run optimizer
options = RadiusOptimizerOptions(
    normalize=True, normalize_metric="surface_area", remove_overlaps=True, verbose=True
)
optimizer = RadiusOptimizer(morph, mm.mesh, options=options)
optimized_graph = optimizer.optimize()  # Returns MorphologyGraph

# save normalized to file
swc_filepath = (
    f"{swc_out_dir}/TS{spine_idx}_s{max_edge_length}_{radius_strategy}_SAnormalized.swc"
)
optimized_graph.to_swc_file(swc_filepath)

# load and plot
swc_model = SWCModel.from_swc_file(swc_filepath)
swc_model.print_attributes(node_info=False, edge_info=False)
frusta = FrustaSet.from_swc_model(swc_model)
title = f"TS{spine_idx}_s{max_edge_length}_{radius_strategy}"
fig = plot_model(swc_model=swc_model, frusta=frusta, slider=True, title=title)
fig.show()
```

```python
validator = Validation(mm, skeleton, morph)
for account_for_overlaps in [False, True]:
    vol_result = validator.compare_volumes(remove_overlaps=account_for_overlaps)
    area_result = validator.compare_surface_areas(remove_overlaps=account_for_overlaps)
    print(f"Validation Results ({radius_strategy}, remove_overlaps={account_for_overlaps}):")
    print("\nVolume Comparison:")
    print(f"  Mesh volume:       {vol_result['mesh_volume']:.4f}")
    print(f"  Morphology volume: {vol_result['morphology_volume']:.4f}")
    print(f"  Ratio:             {vol_result['ratio']:.4f}")
    print(f"  Error:             {vol_result['error']:.4f}")
    print(f"  Relative error:    {vol_result['relative_error']:.2%}")
    print("\nSurface Area Comparison:")
    print(f"  Mesh area:         {area_result['mesh_area']:.4f}")
    print(f"  Morphology area:   {area_result['morphology_area']:.4f}")
    print(f"  Ratio:             {area_result['ratio']:.4f}")
    print(f"  Error:             {area_result['error']:.4f}")
    print(f"  Relative error:    {area_result['relative_error']:.2%}")
```
